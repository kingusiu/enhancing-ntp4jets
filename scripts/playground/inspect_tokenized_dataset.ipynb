# Inspect tokenized JetClass dataset

Look at the parquet schema, row-group layout, and column structure of an
example tokenized JetClass file.

```python
import awkward as ak

fp = (
    "/srv/beegfs/scratch/groups/rodem/lumea/datasets/JetClass/Pythia/cont_alfa_tokenized_allfeat/2026-08-04_14-03-40_gpu024_EquestrianSymbol/train_100M/ZJetsToNuNu_000_tokenized.parquet"
)
```

```python
meta = ak.metadata_from_parquet(fp)
print("num_rows:", meta["num_rows"])
print("num_row_groups:", meta["num_row_groups"])
print("col_counts:", meta["col_counts"])
```

```python
form = meta["form"]
print("top-level fields:", form.fields)
```

```python
arr = ak.from_parquet(fp, row_groups={0})
print(arr.type)
```
